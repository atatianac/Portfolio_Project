```python
zip_path = "../data/raw/C2024_A.zip"

with zipfile.ZipFile(zip_path, "r") as z:
    completions = pd.read_csv(
        z.open("c2024_a.csv"),
        encoding="latin1"
    )

print(completions.shape)
completions.head()
```

```python
completions.columns = completions.columns.str.replace("ï»¿", "", regex=False)
print(completions.columns.tolist())
```

```python
completions_small = completions[
    [
        "UNITID",
        "CIPCODE",
        "AWLEVEL",
        "CTOTALT"
    ]
].copy()

uc_csu_completions = completions_small.merge(
    uc_csu[
        [
            "UNITID",
            "INSTNM",
            "COUNTYNM",
            "LATITUDE",
            "LONGITUD"
        ]
    ],
    on="UNITID",
    how="inner"
)


print(uc_csu_completions.shape)

uc_csu_completions.head(10)
```

```python
uc_csu_completions["INSTNM"].nunique()
```

```python
uc_csu_completions["AWLEVEL"].value_counts().sort_index()
```

```python
degree_completions = uc_csu_completions[
    uc_csu_completions["AWLEVEL"].isin([5, 7, 17, 18])
].copy()

award_labels = {
    5: "Bachelor's degree",
    7: "Master's degree",
    17: "Doctor's - research/scholarship",
    18: "Doctor's - professional practice"
}

degree_completions["AWARD_NAME"] = (
    degree_completions["AWLEVEL"].map(award_labels)
)

degree_completions[
    ["INSTNM", "CIPCODE", "AWARD_NAME", "CTOTALT"]
].head(20)
```

```python
degree_completions.groupby("AWARD_NAME")["CTOTALT"].sum()
```

```python
cip = pd.read_csv("../data/raw/CIPCODE2020.csv")

print(cip.shape)
print(cip.columns.tolist())
cip.head()
```

```python
cip["CIPCode"] = (
    cip["CIPCode"]
    .str.replace('="', '', regex=False)
    .str.replace('"', '', regex=False)
)

print(cip["CIPCode"].head(10).tolist())
```

```python
degree_completions = degree_completions.drop(
    columns=["CIPCode_x", "CIPTitle_x", "CIPCode_y", "CIPTitle_y"]
)

print([col for col in degree_completions.columns if "CIP" in col])
```

```python
degree_completions = degree_completions.merge(
    cip[["CIPCode", "CIPTitle"]],
    left_on="CIPCODE",
    right_on="CIPCode",
    how="left"
)

print("Unmatched CIP codes:")
print(degree_completions["CIPTitle"].isna().sum())
```

```python
unmatched_cips = degree_completions.loc[
    degree_completions["CIPTitle"].isna(),
    "CIPCODE"
].unique()

print(unmatched_cips)
```

```python
degree_completions.loc[
    degree_completions["CIPCODE"] == "99.0000",
    ["INSTNM", "CIPCODE", "AWARD_NAME", "CTOTALT"]
].head(20)
```

```python
degree_completions = degree_completions[
    degree_completions["CIPCODE"] != "99.0000"
].copy()

print(degree_completions["CIPTitle"].isna().sum())
```

```python
top_programs = (
    degree_completions
    .groupby(["CIPCODE", "CIPTitle"], as_index=False)["CTOTALT"]
    .sum()
    .sort_values("CTOTALT", ascending=False)
    .head(20)
)

top_programs
```
